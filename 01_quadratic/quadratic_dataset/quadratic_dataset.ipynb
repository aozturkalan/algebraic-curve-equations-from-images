```python
import numpy as np
import matplotlib.pyplot as plt
import csv
import os

# =====================================================
# Create output directory
# =====================================================

os.makedirs("images", exist_ok=True)

# =====================================================
# Generate dataset
# =====================================================

N_SAMPLES = 5000
saved_count = 0

with open("labels.csv", mode="w", newline="") as f:

    writer = csv.writer(f)
    writer.writerow(["filename", "a", "b", "c", "d", "e", "f"])

    while saved_count < N_SAMPLES:

        # -------------------------------------------------
        # Random quadratic coefficients
        # -------------------------------------------------

        a, b, c = np.random.uniform(-1, 1, 3)
        d, e = np.random.uniform(-2, 2, 2)
        f0 = np.random.uniform(-3, 3)

        # Reject nearly degenerate quadratic part
        if abs(a) + abs(b) + abs(c) < 0.25:
            continue

        # -------------------------------------------------
        # Normalize coefficients
        # -------------------------------------------------

        norm = np.sqrt(
            a*a + b*b + c*c +
            d*d + e*e + f0*f0
        )

        a /= norm
        b /= norm
        c /= norm
        d /= norm
        e /= norm
        f0 /= norm

        # -------------------------------------------------
        # Evaluate on grid
        # -------------------------------------------------

        x = np.linspace(-10, 10, 400)
        y = np.linspace(-10, 10, 400)
        X, Y = np.meshgrid(x, y)

        Z = (
            a * X**2 +
            b * X * Y +
            c * Y**2 +
            d * X +
            e * Y +
            f0
        )

        # Keep only visible curves
        if not (np.any(Z > 0) and np.any(Z < 0)):
            continue

        # -------------------------------------------------
        # Render image
        # -------------------------------------------------

        fig, ax = plt.subplots(figsize=(3, 3))

        ax.contour(
            X,
            Y,
            Z,
            levels=[0],
            colors="black"
        )

        ax.axis("off")
        ax.set_aspect("equal")

        filename = f"sample_{saved_count:04d}.png"

        plt.savefig(
            os.path.join("images", filename),
            dpi=100,
            bbox_inches="tight",
            pad_inches=0
        )

        plt.close(fig)

        # -------------------------------------------------
        # Save coefficients
        # -------------------------------------------------

        writer.writerow([
            filename,
            a,
            b,
            c,
            d,
            e,
            f0
        ])

        saved_count += 1

        if saved_count % 500 == 0:
            print(f"{saved_count}/{N_SAMPLES} samples generated.")

print("✅ Quadratic dataset generation completed.")
```
